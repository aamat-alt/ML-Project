```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Open and load the dataset FIRST
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')

# 2. Set up the visual style for the graphs
sns.set_theme(style="whitegrid")
plt.figure(figsize=(15, 5))

# Graph 1: Bar chart of Churn (Target Variable Distribution)
plt.subplot(1, 3, 1)
sns.countplot(data=df, x='Churn')
plt.title('1. Customer Churn Distribution')

# Graph 2: Histogram of Tenure
plt.subplot(1, 3, 2)
sns.histplot(data=df, x='tenure', bins=30, kde=True)
plt.title('2. Distribution of Tenure (Months)')

# Graph 3: Boxplot of Monthly Charges by Churn
plt.subplot(1, 3, 3)
sns.boxplot(data=df, x='Churn', y='MonthlyCharges')
plt.title('3. Monthly Charges vs Churn')

# Show the graphs
plt.tight_layout()
plt.show()
```
